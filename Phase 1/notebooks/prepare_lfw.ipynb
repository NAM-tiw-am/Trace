```python
import os
import shutil

# ---- EDIT THIS to match the path that worked in debug_lfw.py ----
SOURCE_DIR = r"C:\Users\nt458\OneDrive\Documents\vit documents\semister 3\project exbhition\missing-person-identification\Phase 1\data\lfw-deepfunneled"
DEST_DIR = r"C:\Users\nt458\OneDrive\Documents\vit documents\semister 3\project exbhition\missing-person-identification\Phase 1\data\test"
# -------------------------------------------------------------------

MIN_IMAGES = 3
NUM_PEOPLE = 500  # how many people you want for your test set

os.makedirs(DEST_DIR, exist_ok=True)

# Find people with at least MIN_IMAGES photos
candidates = []
for person_name in sorted(os.listdir(SOURCE_DIR)):
    person_path = os.path.join(SOURCE_DIR, person_name)
    if not os.path.isdir(person_path):
        continue
    images = [f for f in os.listdir(person_path) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    if len(images) >= MIN_IMAGES:
        candidates.append((person_name, images))

print(f"Found {len(candidates)} people with >= {MIN_IMAGES} images.")

# Take the first NUM_PEOPLE candidates
selected = candidates[:NUM_PEOPLE]

for idx, (person_name, images) in enumerate(selected, start=1):
    dest_person_dir = os.path.join(DEST_DIR, f"person_{idx:02d}")
    os.makedirs(dest_person_dir, exist_ok=True)

    for img_idx, img_name in enumerate(sorted(images)[:MIN_IMAGES], start=1):
        src = os.path.join(SOURCE_DIR, person_name, img_name)
        dst = os.path.join(dest_person_dir, f"image_{img_idx:02d}.jpg")
        shutil.copy(src, dst)

    print(f"person_{idx:02d} <- {person_name} ({MIN_IMAGES} images copied)")

print("Done.")
```
